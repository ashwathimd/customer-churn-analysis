```python
df["Total Charges"] = pd.to_numeric(
    df["Total Charges"],
    errors = "coerce"
)
```

```python
df["Total Charges"].isna().sum()
```

### Total Charges

- Converted `Total Charges` from `object` to `float64`.
- Blank values were converted to `NaN`.
- These correspond to customers with zero tenure and are considered valid missing values.
- No imputation was performed.

```python
df.nunique().sort_values()
```

```python
df_clean = df.drop(
    columns=[
        "Count",
        "Country",
        "State",
        "Lat Long"
    ]
)
```

```python
df_clean.shape
```

```python
df_clean.info()
```

## Data Cleaning Summary

### Cleaning Performed

- Converted `Total Charges` from `object` to `float64`.
- Converted blank strings in `Total Charges` to `NaN`.
- Retained the missing values, as they represent customers with zero tenure.
- Removed redundant columns:
  - `Count`
  - `Country`
  - `State`
  - `Lat Long`

### Dataset Status

- Duplicate records: None
- Remaining missing values:
- `Total Charges` (11 valid missing values)
- `Churn Reason` (expected for non-churned customers)

## Data Integrity Validation

- Dataset dimensions were preserved after cleaning.
- No customer records were removed.
- Only valid business-related missing values remain.
- The dataset is consistent and ready for exploratory analysis.

The dataset is now ready for Exploratory Data Analysis (EDA).
